#ColumnTransformer


## 🔹 What is ColumnTransformer?

**`ColumnTransformer`** is a **scikit-learn tool** used to **apply different preprocessing steps to different columns** of a dataset **at the same time**.

👉 Because in real datasets:

* Some columns are **categorical** → need **One-Hot Encoding**
* Some columns are **numerical** → need **Scaling**
* Some columns should be **ignored**

---

## ❓ Why Do We Need ColumnTransformer?

ML models need **numerical input**, but datasets usually look like this:

```
City        Age   Salary
Karachi     25    50000
Lahore      30    60000
Islamabad   28    55000
```

* `City` → categorical → OneHotEncoding
* `Age`, `Salary` → numerical → Scaling

❌ We cannot apply **OneHotEncoder to Age**
❌ We cannot apply **Scaler to City**

✅ **ColumnTransformer solves this problem**

---

## 📌 Basic Syntax

```python
from sklearn.compose import ColumnTransformer
```

```python
ColumnTransformer(
    transformers=[
        ('name', transformer, columns)
    ],
    remainder='drop' or 'passthrough'
)
```

---

## 🧠 Simple Example

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
ct = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first'), ['City']),
        ('num', StandardScaler(), ['Age', 'Salary'])
    ],
    remainder='passthrough'
)
```

### Explanation:

* `'cat'` → name (anything you like)
* `OneHotEncoder()` → applied to **City**
* `StandardScaler()` → applied to **Age & Salary**
* `remainder='passthrough'` → keeps remaining columns (if any)

---

## 🧪 How It Works Internally

### Before

```
City        Age   Salary
Karachi     25    50000
Lahore      30    60000
```

### After

```
City_Lahore  Age_scaled  Salary_scaled
0            -1.0        -1.0
1             1.0         1.0
```

---

## 📌 Using ColumnTransformer in ML Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
```

```python
pipe = Pipeline(steps=[
    ('preprocessing', ct),
    ('model', LinearRegression())
])
```

```python
pipe.fit(X_train, y_train)
```

✔ Clean
✔ No data leakage
✔ Production-ready

---

## ⚠️ `remainder` Parameter

| Value           | Meaning                       |
| --------------- | ----------------------------- |
| `'drop'`        | Drop unused columns (default) |
| `'passthrough'` | Keep remaining columns        |

#Column Transformer

````python
from sklearn.compose import ColumnTransformer
````

````python
transformer = ColumnTransformer(transformers=[
('tnf1',MinMaxScaler(),['fever']),
('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
````

````python
transformer.fit_transform(x_train)
````

````python
transformer.fit_transform(x_test)
````
