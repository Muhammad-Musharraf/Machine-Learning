#3.OneHotEncoding

## 🔹 What is One-Hot Encoding?

**One-Hot Encoding (OHE)** is a **data preprocessing technique** used in **Machine Learning** to convert **categorical (text) data** into **numerical (binary) form**, so that ML models can understand it.

---

## ❓ Why Do We Need One-Hot Encoding?

Most machine learning algorithms **cannot work with text** like:

```
Color = Red, Blue, Green
City = Karachi, Lahore, Islamabad
```

They only understand **numbers**.

👉 One-Hot Encoding solves this problem by creating **binary (0/1) columns** for each category.

---

## 📌 Simple Example

### Original Data

```
Color
Red
Blue
Green
```

### After One-Hot Encoding

```
Red   Blue   Green
1      0       0
0      1       0
0      0       1
```

✔ Each category gets its own column
✔ Only one column has `1` → that’s why it’s called **one-hot**

---

## 🧠 Why It Is Called “One-Hot”?

Because:

* **One value = 1 (hot)**
* **All others = 0 (cold)**

Example:

```
Blue → [0, 1, 0]
```

---

## 🚀 Real-World Example (ML)

### Dataset

```
City        Price
Karachi     50
Lahore      45
Islamabad   60
```

### After One-Hot Encoding

```
Karachi  Lahore  Islamabad  Price
1        0       0           50
0        1       0           45
0        0       1           60
```

Now ML models can **learn patterns correctly**.

---

## ⚠️ Dummy Variable Trap (Important)

If we create **all columns**, one column can be predicted from others → causes **multicollinearity**.

### Solution:

Drop one column

Example:

```
Karachi  Lahore
1        0   → Islamabad
0        1   → Lahore
0        0   → Karachi
```

This is why we often use:

```python
OneHotEncoder(drop='first')
```

---

## 🆚 One-Hot Encoding vs Label Encoding

| Feature       | One-Hot Encoding | Label Encoding            |
| ------------- | ---------------- | ------------------------- |
| Best for      | Nominal data     | Ordinal data              |
| Order matters | ❌ No             | ✅ Yes                     |
| Example       | City, Color      | Grade (Low, Medium, High) |

---

## ✅ When to Use One-Hot Encoding?

✔ Nominal categorical data
✔ Linear Regression
✔ Logistic Regression
✔ Neural Networks

❌ Not ideal for:

* Very high-cardinality features (many categories)
* Tree-based models (Random Forest, Decision Tree)

---
